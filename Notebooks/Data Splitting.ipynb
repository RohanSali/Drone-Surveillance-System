```python
import os
import random
import shutil

def move(source_folder,destination_folder,number_of_images_to_move):
    # === SUPPORTED IMAGE EXTENSIONS ===
    image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.gif')
    
    # === FETCH AND FILTER IMAGE FILES ===
    all_images = [f for f in os.listdir(source_folder) if f.lower().endswith(image_extensions)]
    
    # === HANDLE IF FEWER IMAGES THAN REQUESTED ===
    if len(all_images) < number_of_images_to_move:
        print(f"Only {len(all_images)} images found. Moving all.")
        number_of_images_to_move = len(all_images)
    
    # === RANDOM SELECTION ===
    selected_images = random.sample(all_images, number_of_images_to_move)
    
    # === MOVE FILES ===
    for image in selected_images:
        src_path = os.path.join(source_folder, image)
        dst_path = os.path.join(destination_folder, image)
        shutil.move(src_path, dst_path)
    
    print(f"{number_of_images_to_move} image(s) moved from '{source_folder}' to '{destination_folder}'.")
```

```python
source_folder = 'dataset_casualities/accident'
destination_folder = 'dataset_casualities/train/accident'
number_of_images_to_move = 15000
move(source_folder,destination_folder,number_of_images_to_move)
```
