Importing Liberaries

```python
import torch
from facenet_pytorch import InceptionResnetV1, MTCNN
from PIL import Image

# Load pretrained FaceNet model
model = InceptionResnetV1(pretrained='vggface2').eval()  # Or 'casia-webface' for alternative
mtcnn = MTCNN(image_size=160, margin=0)  # For face detection and alignment
```

```python
def get_embedding(image_path):
    img = Image.open(image_path)
    face = mtcnn(img)
    if face is not None:
        return model(face.unsqueeze(0))
    return None

# Example: Compare two images
emb1 = get_embedding('/content/img1.jpg')
emb2 = get_embedding('/content/img2.jpg')

if emb1 is not None and emb2 is not None:
    cosine_sim = torch.nn.functional.cosine_similarity(emb1, emb2).item()
    threshold = 0.6  # Adjust based on testing
    if cosine_sim > threshold:
        print("Faces match! Similarity:", cosine_sim)
    else:
        print("Faces do not match. Similarity:", cosine_sim)
```
